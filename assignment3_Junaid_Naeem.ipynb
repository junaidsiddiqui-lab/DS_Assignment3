```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Airbnb dataset
df = pd.read_csv('/content/drive/MyDrive/Airbnb_Data.csv')

# Create a dataset for EDA
eda_df = df.copy()

print("Dataset shape:", eda_df.shape)
print("\nColumns:")
print(eda_df.columns.tolist())

display(eda_df.head())
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=eda_df, x="log_price", bins=40, kde=True)

plt.title("Distribution of Airbnb Listing Prices (Log Scale)")
plt.xlabel("Log Price")
plt.ylabel("Number of Listings")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=eda_df,
    x="room_type",
    order=eda_df["room_type"].value_counts().index
)

plt.title("Distribution of Airbnb Room Types")
plt.xlabel("Room Type")
plt.ylabel("Number of Listings")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### Interpretation: Distribution of Room Types

The bar chart shows the distribution of Airbnb listings by room type. **Entire home/apartment** is the most common category, with approximately 41,000 listings. **Private rooms** are the second most common, with around 30,500 listings. **Shared rooms** are the least common, with approximately 2,300 listings. This shows that most Airbnb listings in the dataset offer an entire home or apartment, while shared rooms are much less common.

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=eda_df, x="room_type", y="log_price")

plt.title("Airbnb Prices by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Log Price")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### Interpretation: Airbnb Prices by Room Type

The box plot compares Airbnb listing prices across three room types. Entire homes/apartments have the highest median log price, at approximately 5.1. Private rooms have a lower median of around 4.3, while shared rooms have the lowest median of approximately 3.8. This suggests that entire homes and apartments are generally more expensive than private and shared rooms. The graph also shows several outliers, representing listings with unusually high or low prices. One shared-room listing has a log price of zero, which may require further investigation.

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=eda_df,
    x="accommodates",
    y="log_price",
    alpha=0.3
)

plt.title("Relationship Between Price and Accommodation Capacity")
plt.xlabel("Number of Guests Accommodated")
plt.ylabel("Log Price")
plt.tight_layout()
plt.show()
```

### Interpretation: Price vs. Accommodation Capacity

The scatter plot examines the relationship between the number of guests a property can accommodate and its log price. A general upward trend would indicate that properties accommodating more guests tend to have higher prices. The spread of points also shows that properties with similar capacities can have different prices. This suggests that accommodation capacity is an important factor, but it is not the only factor affecting listing prices.

```python
plt.figure(figsize=(11, 6))

top_properties = eda_df["property_type"].value_counts().head(10).index

sns.countplot(
    data=eda_df[eda_df["property_type"].isin(top_properties)],
    y="property_type",
    order=top_properties
)

plt.title("Top 10 Airbnb Property Types")
plt.xlabel("Number of Listings")
plt.ylabel("Property Type")
plt.tight_layout()
plt.show()
```

### Interpretation: Property Type Distribution

This bar chart displays the ten most common property types in the dataset. It helps identify which kinds of properties are most frequently listed on Airbnb. Property types with longer bars have more listings, while shorter bars represent less common types. ### Handling Missing Values

```python
# Detect potential outliers using IQR
for col in [
    "log_price",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "number_of_reviews",
    "review_scores_rating"
]:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ).sum()

    print(f"{col}: {count} potential outliers")
```

```python
# Create a separate copy for preprocessing
df_preprocessed = df_clean.copy()

# Remove the listing ID from potential model features
df_preprocessed = df_preprocessed.drop(columns=["id"])

print("Cleaned dataset shape:", df_clean.shape)
print("Preprocessing dataset shape:", df_preprocessed.shape)
```

```python
# Create useful features from dates
df_preprocessed["host_since_year"] = (
    df_preprocessed["host_since"].dt.year
)

df_preprocessed["first_review_year"] = (
    df_preprocessed["first_review"].dt.year
)

df_preprocessed["last_review_year"] = (
    df_preprocessed["last_review"].dt.year
)

# Count amenities listed for each property
def count_amenities(value):
    if pd.isna(value) or value == "Unknown" or value.strip() == "":
        return 0
    return len(value.strip("{}").split(","))

df_preprocessed["amenities_count"] = (
    df_preprocessed["amenities"].apply(count_amenities)
)

# Check the new features
print(df_preprocessed[
    [
        "host_since_year",
        "first_review_year",
        "last_review_year",
        "amenities_count"
    ].head()
])

print("\nNew dataset shape:", df_preprocessed.shape)
```

```python
# Display the first 5 rows of the new features
print(df_preprocessed[
    [
        "host_since_year",
        "first_review_year",
        "last_review_year",
        "amenities_count"
    ]
].head())

print("\nNew dataset shape:", df_preprocessed.shape)
```

```python
# Convert percentage text into numeric proportions
df_preprocessed["host_response_rate"] = (
    df_preprocessed["host_response_rate"]
    .replace("Unknown", np.nan)
    .str.rstrip("%")
    .astype(float) / 100
)

# Check the conversion
print(df_preprocessed["host_response_rate"].head(10))
print("\nMissing response rates:",
      df_preprocessed["host_response_rate"].isnull().sum())
```

```python
# Check missing values in the preprocessing dataset
missing_counts = df_preprocessed.isnull().sum()
print(missing_counts[missing_counts > 0].sort_values(ascending=False))
```

```python
# Record which values were missing before filling them
df_preprocessed["response_rate_missing"] = (
    df_preprocessed["host_response_rate"].isna().astype(int)
)

df_preprocessed["rating_missing"] = (
    df_preprocessed["review_scores_rating"].isna().astype(int)
)

df_preprocessed["first_review_missing"] = (
    df_preprocessed["first_review"].isna().astype(int)
)

df_preprocessed["last_review_missing"] = (
    df_preprocessed["last_review"].isna().astype(int)
)

df_preprocessed["host_since_missing"] = (
    df_preprocessed["host_since"].isna().astype(int)
)

# Fill missing numeric values with their medians
numeric_fill_cols = [
    "host_response_rate",
    "review_scores_rating",
    "host_since_year",
    "first_review_year",
    "last_review_year"
]

for col in numeric_fill_cols:
    df_preprocessed[col] = df_preprocessed[col].fillna(
        df_preprocessed[col].median()
    )

# Verify the selected columns
print(df_preprocessed[numeric_fill_cols].isnull().sum())
print("\nNew dataset shape:", df_preprocessed.shape)
```

### Categorical Encoding

One-hot encoding was selected for categorical features such as property type, room type, bed type, cancellation policy, and city. This creates separate binary columns for their categories without assigning artificial numerical rankings. Free-text fields and high-cardinality fields, such as listing names, descriptions, amenities, and URLs, were excluded from this basic encoding step because they require specialized processing.

```python
import pandas as pd

# Create a separate copy for encoding
df_encoded = df_preprocessed.copy()

# Select categorical columns suitable for one-hot encoding
categorical_cols = [
    "property_type",
    "room_type",
    "bed_type",
    "cancellation_policy",
    "city",
    "host_has_profile_pic",
    "host_identity_verified"
]

# Convert selected categorical columns into binary columns
df_encoded = pd.get_dummies(
    df_encoded,
    columns=categorical_cols,
    dtype=int
)

print("Encoded dataset shape:", df_encoded.shape)
print(df_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Create another copy for scaling
df_scaled = df_encoded.copy()

# Select continuous numeric features to scale
scale_cols = [
    "log_price",
    "accommodates",
    "bathrooms",
    "bedrooms",
    "beds",
    "number_of_reviews",
    "review_scores_rating",
    "latitude",
    "longitude",
    "host_response_rate",
    "host_since_year",
    "first_review_year",
    "last_review_year",
    "amenities_count"
]

# Scale only the selected numeric columns
scaler = StandardScaler()
df_scaled[scale_cols] = scaler.fit_transform(df_scaled[scale_cols])

print("Scaled dataset shape:", df_scaled.shape)
print(df_scaled[scale_cols].describe().round(2))
```

### Numerical Scaling

Standardization was applied to selected numerical features using `StandardScaler`. The transformation places the features on comparable scales, with values centered around zero and scaled to approximately unit standard deviation. Binary indicators and one-hot encoded columns were not scaled. For future predictive modeling, the scaler must be fitted only on the training data and then applied to validation and test data to prevent data leakage.

```python
# Check skewness of selected numeric features
skew_cols = [
    "log_price",
    "accommodates",
    "number_of_reviews",
    "beds",
    "amenities_count"
]

print(df_preprocessed[skew_cols].skew().round(2))
```

```python
# Create a log-transformed review-count feature
df_preprocessed["number_of_reviews_log"] = np.log1p(
    df_preprocessed["number_of_reviews"]
)

# Compare the original and transformed distributions
print("Original skewness:",
      round(df_preprocessed["number_of_reviews"].skew(), 2))

print("Transformed skewness:",
      round(df_preprocessed["number_of_reviews_log"].skew(), 2))
```

### Skewness and Data Transformation

Skewness was examined for selected numerical features. A logarithmic transformation using `log1p()` was applied to the number of reviews to reduce the effect of very large review counts. The original feature was retained for comparison. The `log_price` feature was not transformed again because it was already log-transformed in the original dataset. Transformations should be evaluated before their use in a predictive model.
